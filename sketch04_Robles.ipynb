```python
DATA_BASE = "https://raw.githubusercontent.com/pjbitterman/KSU_x9076/main/data"

import pandas as pd

ohio_precip = pd.read_csv(f"{DATA_BASE}/part1/ohio_precip_2025.csv")
ohio_precip.head()
```

```python
ohio_precip["precip_mm_2025"].describe()
```

```python
wettest_stations = ohio_precip.sort_values("precip_mm_2025", ascending=False)
wettest_stations.head()
```

```python
final_table = wettest_stations.head()
final_table[["name", "precip_mm_2025"]]
```

```python
final_table = wettest_stations.head().copy()
final_table["mean"] = ohio_precip["precip_mm_2025"].mean()
final_table[["name", "precip_mm_2025", "mean"]]
```

```python
wettest_station = final_table.iloc[0]
difference = wettest_station["precip_mm_2025"] - wettest_station["mean"]
print(round(difference, 2))
```

The wettest station "CINCINNATI MUNI AP LUNKEN FLD" is 486.84 mm above the mean precipitation.
It doesn't surprise me, the station is next to the Ohio River valley so it gets a lot of moisture from it and this exposure tends to make the air of a place more humid, encreasing the probability of rain.
